# Tables and Figures Summary
## LinkedIn Entrepreneurship & Experience [Paper 2]
**Date:** July 2026

This document provides a comprehensive summary of all tables and figures from the research paper, highlighting key findings, regression results, and robustness checks. It starts with the main tables and figures prior to the appendix, followed by the appendix tables and figures.

## Executive Summary

The paper investigates the relationship between individuals' career backgrounds (specifically **Business** vs. **Technical** roles) and their likelihood of transitioning into entrepreneurship, as well as the success (growth size) of their subsequent ventures.

### Key Findings:
1. **Founding Likelihood:** Individuals in **Business roles** are significantly more likely to become entrepreneurs when transitioning to a subsequent role compared to those in **Technical roles** (baseline). Regression results show a consistent positive coefficient for Business roles across all specifications.
2. **Education Interactions:** The interaction between holding a **Business role** and a **STEM degree** is positive and highly significant, suggesting that the combination of business experience and technical education is a strong predictor of entrepreneurship.
3. **High-Growth Ventures:** Within the sample of founders, individuals leaving **Business roles** are significantly more likely to found a **High-Growth venture** (defined as employing more than 500 individuals) than those from technical roles.
4. **Career Age & Demographics:** Business founders tend to start their ventures with slightly more career experience (career age mean of 12.2 years) compared to technical founders (mean of 10.8 years).

* **Key Takeaway:** Business backgrounds dominate among founders leaving Consulting (~75%) and Finance (~65%), whereas Technical backgrounds are relatively more common among founders leaving Health/Biotech (~35%) and Manufacturing/Defense (~25%).

### Figure 2. Previous Role of Founders (Business vs. Technical) by Size of Company Founded

![Figure 2](figures/Figure_2.png)

* **Description:** Analyzes the proportion of Silicon Valley-based technology founders from business vs. technical roles, stratified by the final size of their venture (<10, 10-49, 50-249, and 250+ employees).
* **Key Takeaway:** As the size of the founded venture increases (which serves as a proxy for growth and success), the share of founders coming from **Business roles** increases dramatically (from ~44% to over 60%), while the share from **Technical roles** remains stable or slightly declines. This suggests that business experience prior to founding is strongly associated with scaling a startup.

### Appendix A. Detailed Information About Data From Figures

Appendix A contains detailed tables detailing the shares of roles, industries, company sizes, and degrees of founders.

*   **Table A.A.1 (Share of Prior Roles):** Outlines raw percentages corresponding to Figure 1a. Business roles are the most common prior background, standing at 48.9% for the US-based tech founders sample.
*   **Table A.A.2 (Examples of Roles):** Lists typical titles for role clusters (e.g., *Business:* Product Manager, Consultant, Director; *Technical:* Software Engineer, Systems Architect, Scientist).
*   **Table A.B.1 & Table A.B.2 (Parent Firms & Industries):** Details industry distributions. Out of technology founders, about 32% come from technical roles and 53% from business roles, which aligns closely with the baseline employment ratios of their parent industries.
*   **Table A.C.1 (Role by Company Size):** Details how role distributions shift as the size of the founded venture grows, matching the graphical trend shown in Figure 2.
*   **Table A.D.1 & Table A.D.2 (Education details):** Tabulates degrees. Computer Science and Business Administration are the leading fields of study.
*   **Table A.E.1 (Prior Roles by Gender):** Details how prior role distributions differ between male and female founders.
